# Funzioni matematiche e numeri random

### Valore assoluto

```python
x = -3
result = abs(x)
print("Il valore assoluto di x: ", result)
```

### Arrotondamento

```python
x = 2.55554
# Arrotondamento bancario
result = round(x)
print("Valore arrotondato di x:", result)
print(f"Valore arrotondato di x: {x:.0f}")
```

## Funzioni matematiche avanzate con math

```python
import math
```

### Arrotondamento per difetto

```python
# Arrotondamento per difetto forzato
x = 4.8
# floor va verso -∞
result = math.floor(x)
print("Valore arrotondato per difetto di x: ", result)
```

### Arrotondamento per eccesso

```python
x = 4.01
# ceil va verso +∞
result = math.ceil(x)
print("Valore arrotondato per eccesso di x: ", result)
```

### Troncamento

```python
x= -4.5
# trunc va verso 0
result = math.trunc(x)
print("Valore troncato di x: ", result)
```

### Radice quadrata

```python
x = 16
result = math.sqrt(x)
print ("Il valore della radice quadrata di x:", result)
```

### Funzioni trigonometriche

```python
x = 45
x_rad = math.radians(x)

x_sin = math.sin(x_rad)
print("Valore del seno di x: ", x_sin)
x_cos = math.cos(x_rad)
print("Valore del coseno di x: ", x_cos)
```

### Costanti matematiche

```python
print("Pi greco: ", math.pi)
print("Numero di Nepero: ", math.e)
```

## Numeri random

```python
import random
```

### Numeri interi casuali

```python
dice = random.randint(1,6)
print("Valore del dado: ", dice)
```

### Numeri decimali casuali

```python
# numero casuale tra 0 e 1
x = random.random()
print(x)
```

```python
x = random.uniform(10.0, 20.0)
print(x)
```
