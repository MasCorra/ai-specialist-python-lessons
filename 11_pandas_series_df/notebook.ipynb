## Accesso agli elementi
.loc[]--> seleziona elemento di quell'indice

.iloc[] --> stessa cosa con le posizioni numeriche

```python
prezzi_prodotti = pd.Series([3,4,5,6,7], index = ['PC','Telefono', 'Tastiera', 'Mouse' ,'Cuffie'], name = 'Prodotti')
print("Prezzo Mouse:", prezzi_prodotti.loc['Mouse'])
print(f'prezzo della tastiera e del mouse\n{prezzi_prodotti.loc[['Tastiera', 'Mouse']]}')
```

```python
print('Prezzo Primo Elemento', prezzi_prodotti.iloc[0])
print('Prezzo Ultimo Elemento', prezzi_prodotti.iloc[-1])
```

## Slicing
.loc[inizio:fine] posizione di fine è inclusa, si puo mettere anche il passo [inizio:fine:passo]
.iloc[inzio:fine] posizione di fine è esclusa

```python
#Stampare il prezzo dei primi tre
print("Prezzi dei primi 3 prodotti: ")
print(prezzi_prodotti.iloc[0:3])
```

```python
#Stampare il prezzo dei primi tre
print("Prezzi dei primi 4 alternati: ")
print(prezzi_prodotti.iloc[0:4:2])
```

```python
#.loc[]
print("Tutti quelli compresi tra pc e tastiera")
print(prezzi_prodotti.loc['PC':'Tastiera'])
```

## Modifica valori
le Series sono mutabili: possiamo modificare, aggiungere togliere valori
generalmente si crea una copia .copy() della serie originale per fare i controlli

```python
prezzi_aggiornati = prezzi_prodotti.copy()
print(prezzi_aggiornati)
prezzi_aggiornati['Telefono'] = 24
print(prezzi_aggiornati)
```

```python
prezzi_aggiornati['Webcam'] = 30
print('I prezzi agigornati sono:')
prezzi_aggiornati
```

```python
senza_telefono = prezzi_aggiornati.drop('Telefono')
print(senza_telefono)
```

## Operazioni Matematiche
operazioni vettoriali: non hanno bisogno di cicli for, nb la stessa operazione di applica a tutti gli elementi della serie (che rispettano le condizioni che volete voi)

```python
prezzi_scontati = prezzi_aggiornati * 0.79
print(prezzi_aggiornati)
print(prezzi_scontati)
```

```python
prezzi_aumentati = prezzi_aggiornati + 5
print(prezzi_aumentati)
```

## Confronto sui valori e booleani

## Filtraggio
si prende la serie di booleani (mask) e selezioniamo i valori della serie originale usando quella maschera

## Codizioni
si possono mettere piu condizioni ( AND & OR | NOT ~)
.between() ( valori compresi tra inizio e fine estremi inclusi e ci restituisce una mask di bool),
.isin() ( ci restituisce una mask di booleani)

```python
#prezzi rea 5 e 10 euro
print(prezzi_aggiornati.loc[(prezzi_aggiornati<=10)&(prezzi_agigornati>=5)]) # stessa cosa con or and not
```

```python
#between()
# prezzi tra 5 e 10 euro
print(prezzi_aggiornati,loc[prezzi_aggiornati.between(5,10)])
```

```python
#.isin()
#voglio solamente gli indici di TELEDFONO e PC
print(prezzi_aggiornati.loc[prezzi_aggiornati.index.isin(['PC', 'Telefono'])])
```

# Valori mancanti
- .isna() --> verifica quali valori sono mancanti
- .notna() --> verifica quali valori non sono nulli
- .count() --> ritorna il numero di valori non nulli
- .size() --> ritorna il numero totale di valori

```python
valutazioni = pd.Series([4.2, None, 5, None, 12], index = ['PC', 'Telefono', 'Cuffie',
                                                           'Mouse','Tastiera'])
print('maschera:')
print(valutazioni.isna())
print('Valutazioni con argomento nullo:')
valori_nulli = valutazioni[valutazioni.isna()]
print(valori_nulli)
print('Valutazione con argomento nullo')
valori_non_nulli = valutazioni[~valutazioni.isna()]
print(valori_non_nulli)
```

```python
print('conta elementi non nulli')
print(valutazioni.count())
print('conta elementi totali')
print(valutazioni.size)
```

# Gestione dei valori mancanti:


```python
copia_valutazioni = valutazioni.copy()
print(copia_valutazioni.dropna()) # elimina tutte le righe con valori mancanti
#voglio sostituire i valori mancanti con la media
media_valutazioni = copia_valutazioni.mean()
print(copia_valutazioni)
```

# Ordinamento
.sort_values() ordina per valore
.sort_index() ordina per indice
posso richiedere ascending=True (crescente) oppure False (decrescente)

```python
print('categorie in ordine alfabetico')
print(categorie.sort_values(ascending=True)) #se non specificato, di default è True

print('oridnamento per indice')
print(categorie.sort_index())
```
