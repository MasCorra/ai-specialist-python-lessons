## Codizioni
si possono mettere piu condizioni ( AND & OR | NOT ~)
.between() ( valori compresi tra inizio e fine estremi inclusi e ci restituisce una mask di bool),
.isin() ( ci restituisce una mask di booleani)

```python
#prezzi rea 5 e 10 euro
print(prezzi_aggiornati.loc[(prezzi_aggiornati<=10)&(prezzi_aggiornati>=5)]) # stessa cosa con or and not
```

```python
#between()
# prezzi tra 5 e 10 euro
print(prezzi_aggiornati.loc[prezzi_aggiornati.between(5,10)])
```

```python
#.isin()
#voglio solamente gli indici di TELEDFONO e PC
print(prezzi_aggiornati.loc[prezzi_aggiornati.index.isin(['PC', 'Telefono'])])
```

# Valori mancanti
- .isna() --> verifica quali valori sono mancanti
- .notna() --> verifica quali valori non sono nulli
- .count() --> ritorna il numero di valori non nulli
- .size() --> ritorna il numero totale di valori

```python
valutazioni = pd.Series([4.2, None, 5, None, 12], index = ['PC', 'Telefono', 'Cuffie',
                                                           'Mouse','Tastiera'])
print('maschera:')
print(valutazioni.isna())
print('Valutazioni con argomento nullo:')
valori_nulli = valutazioni[valutazioni.isna()]
print(valori_nulli)
print('Valutazione con argomento nullo')
valori_non_nulli = valutazioni[~valutazioni.isna()]
print(valori_non_nulli)
```

```python
print('conta elementi non nulli')
print(valutazioni.count())
print('conta elementi totali')
print(valutazioni.size)
```

# Gestione dei valori mancanti:


```python
copia_valutazioni = valutazioni.copy()
print(copia_valutazioni.dropna()) # elimina tutte le righe con valori mancanti
#voglio sostituire i valori mancanti con la media
media_valutazioni = copia_valutazioni.mean()
print(copia_valutazioni)
```

# map() e apply()

```python
df_modificato_copia=df_modificato.copy()
nomi_categorie = {'informatica':'Informatica', 'alimentari':'Alimentari'}
df_modificato_copia['categoria']= df_modificato_copia['categoria'].map(nomi_categorie)
df_modificato_copia

# apply(vostra funzione/lambda) applica la funzione
#vettoriale
df_modificato_copia['prezzo_iva_vettoriale'] = df_modificato_copia['prezzo'] * 1.22
df_modificato_copia['prezzo_iva_lambda'] = df_modificato_copia['prezzo'].apply(lambda x: x*1.22)
```
