```python
class Person:
    def __init__(self, first_name, surname, age): #metodo che viene richiamato quando creo la mia persona
        self.first_name = first_name #memorizza il nome per quell'oggetto
        self.surname = surname
        self.name = first_name + " " + surname
        self.age = age
        self.adult = self.age > 18
    def greet(self):
        print(f"Ciao! sono {self.name} e ho {self.age} anni!")
chiara = Person("Chiara", "Scagliola", 27)
chiara.greet()
```

```python
class BankLog:
    def __init__(self,old_balance,new_balance, operation):
        self.old_balance = old_balance
        self.new_balance = new_balance
        self.operation = operation

    def summary(self):
        print(f"[{self.operation}] {self.old_balance} -> {self.new_balance}")

class BankAccount:
    def __init__(self, first_name, last_name, email):
        self.first_name = first_name
        self.last_name = last_name
        self.email = email
        self.__balance = 0
        self.__logs = []

    def show(self):
        print(f"il conto corrente è di: {self.first_name} {self.last_name}")
        print(f"il saldo è di {self.__balance}")
        for log in self.__logs:
            log.summary()

    def deposit(self, amount):
        old_balance = self.__balance
        self.__balance += amount
        self.__logs.append(BankLog(old_balance, self.__balance, "+"))

    def withdrow(self, amount):
        old_balance = self.__balance
        self.__balance -= amount
        self.__logs.append(BankLog(old_balance, self.__balance, "-"))

account = BankAccount("Chiara","Scagliola","chiara.sca@gmail.com")
account.show()
account.deposit(1000)
account.withdrow(50)
account.show()
```

```python
class BankAccount:
    bank_name = "Intesa San Paolo"
    total_accounts = 0
    def __init__(self, first_name, last_name, email):
        self.first_name = first_name
        self.last_name = last_name
        self.email = email
        self.__balance = 0
        BankAccount.total_accounts += 1

    def show(self):
            print(f"il conto corrente è di: {self.first_name} {self.last_name}")
            print(f"il saldo è di {self.__balance}")

    def deposit(self, amount):
        old_balance = self.__balance
        self.__balance += amount

    def withdrow(self, amount):
        old_balance = self.__balance
        self.__balance -= amount

    @classmethod
    def show_number_accounts(cls):
        print("il numero di account registrati è:", cls.total_accounts)
account1 = BankAccount("Chiara", "Scagliola", "chiara.scagliola@gmail.com")
account1.show()
print("----------------------------------------")
account2 = BankAccount("Lucia", "Ferni", "fernilucia@gmail.com")
account1.show()
account2.show()
print("----------------------------------------")
print(BankAccount.total_accounts)
BankAccount.show_number_accounts()
account1.show_number_accounts()
```
