## Funzioni

```python
def greet(name: str):
    print("Ciao", name)

greet("Luca")

def get_full_name(first_name: str, last_name: str):
    return f"{first_name} {last_name}"
name: str = get_full_name("Franco", "Rossi")
name2 = name + "10"
result: int = get_full_name("Franco", "Rossi")
print (result)
```

## Tipi multipli

```python
def greet(name: str | int):
    print("Ciao", name)
greet ("Luca")
greet(1)
```

```python
def greet(name: str | None = None):
    if name is None:
        print("Ciao sconosciuto!")
    else:
        print("Ciao", name)
greet()
greet("Luca")
```
