```python
lista_nomi= ['Marta', 'Andrea', 'Paola', 'Giuseppe']
print('lunghezza della lista', len(lista_nomi))
print('Marta è nella lista?', 'Marta' in lista_nomi)
print('Angelo è nella lista?', 'Angelo' in lista_nomi)
lista_nomi_2 = ['Laura', 'Martino', 'Lucia', 'Gabriele']
lista_totale=lista_nomi+lista_nomi_2
print('lista totale dei parenti', lista_totale)
print('lista ripetuta', lista_nomi_2*2)
```

```python
numbers=[5,4,7,8,12,55,21]
somma_numeri= sum(numbers)
massimo = max(numbers)
minimo = min(numbers)
print('somma numeri', somma_numeri)
print(f'massimo {massimo}, minimo {minimo}')
```

## Metodi delle liste

```python
lista_spesa = ['pane', 'pasta', 'olio', 'frutta', 'verdura']
print('Lista spesa', lista_spesa)
# aggiungere elemento alla fine della lista
lista_spesa.append('latte')
print('Lista spesa post append',lista_spesa)
lista_spesa.extend(['burro','zucchero'])
print('Lista spesa post extend',lista_spesa)
lista_spesa.insert(1,'parmigiano')
print('Lista spesa post insert',lista_spesa)
```

```python
#per togliere
lista_spesa.remove('pane')
print('Lista spesa post remove',lista_spesa)
#Rimuove di default in ultima posizione
lista_spesa.pop()
print('Lista spesa post pop',lista_spesa)
#Rimuove in prima posizione
lista_spesa.pop(0)
print('Lista spesa post pop2',lista_spesa)
lista_spesa.clear()
print('Lista spesa post clear',lista_spesa)
```

## Ordinare liste

```python
lista_spesa = ['pane', 'pasta', 'olio', 'frutta', 'verdura']
lista_spesa.sort()
print('lista ordinata in ordine ascendente', lista_spesa)
lista_spesa.sort(reverse=True)
print('lista ordinata in ordine discendente', lista_spesa)
```

## Ricerca e Conteggio

```python
lista_spesa = ['pane', 'pasta', 'olio', 'frutta', 'verdura']
indice = lista_spesa.index('pane')
print('indice: ', indice)
lista_spesa= lista_spesa*2
conta = lista_spesa.count('olio')
print('conta: ', conta)
```

## Slicing delle liste
lista[start:stop:passo]

```python
animali= ['Gatto', 'Topo', 'Cane', 'Papera', 'Coniglio']
print('primi 3: ', animali[:3])
print('ultimi 3: ', animali[-3:])
print('tutti con passo 2:', animali[::2])
print('inversione:', animali[::-1])
animali[2:2]=['Paguro', 'Scimmia', 'Avvoltoio']
print(animali)
```
