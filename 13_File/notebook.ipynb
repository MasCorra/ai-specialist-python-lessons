## Lettura e scrittura

## Scrittura

```python
file = open("example.txt", "w")

file.write("Hello, World!")
file.close()
```

```python
file = open("data.csv", "w")
file.write("name,score\n")
file.write("Simone Colomba, 50\n")
file.write("Mario Rossi, 40\n")
file.close()
```

```python
file = open("example.txt", "w")
lines = [
    "Nome 1",
    "Nome 2",
    "Nome 3",
]
file.writelines([line + "\n" for line in lines])
file.close()
```

```python
file = open("example.txt", "a")

file.write("--------------")
file.close()
```

## Lettura

```python
file = open("example.txt", "r")
content = file.read()
file.close()
print(content)
```

```python
file = open("example.txt", "r")

for line in file:
    print(line.strip() , end=" ") #strip agisce sulle righe del file, end= agisce sul print

file.close()
```

```python
file = open("example.txt", "r")

lines = file.readlines()
print(lines)

file.close()
```

### Modalita di apertura dei file
r --> Lettura (read)

w --> Scrittura (write)

a --> Aggiunta (append)

x --> SOLO Creazione, prima scrittura (Exclusive creation)

r+ --> Lettura e scrittura (non sovrascrive)

w+ --> Scrittura e lettura (sovrascrive)

## Gestione del puntatore

```python
file = open("example.txt", "r")

print("Puntatore", file.tell())
print("======")
content = file.read(10)
print(content)
print("======")
print("Puntatore", file.tell())
print("======")
content2 = file.read()
print(content2)
file.close()
```

### Spostare puntatore

```python
file = open("example.txt", "r")

print("Puntatore", file.tell())
print("======")
content = file.read()
print(content)
print("======")
file.seek(10)
print("Puntatore", file.tell())
print("======")
content2 = file.read(5)
print(content2)
print("Puntatore", file.tell())
file.close()
```

## Gestione dei file con Context Manager

```python
with open("example.txt", "r") as file:
    content = file.read()

print(content)
```

```python
with open("example.txt", "r") as file:
    content = file.read()

content += "\n\n=========\nFINE\n==========="
print(content)

with open("example.txt", "w") as file:
    file.write("content")

print(content)
```

## Codifica dei caratteri (encoding)

```python
with open("exampple_utf8.txt", "w", encoding="utf-8") as file:
    file.write("🍕")
```

```python
with open("exampple_utf8.txt", "r", encoding="utf-8") as file:
    content= file.read()
print(content)
```
