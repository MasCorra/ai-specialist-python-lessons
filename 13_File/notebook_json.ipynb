# Gestione di dati JSON in Python

## Importazione del modulo JSON

```python
import json
```

### Conversione da dizionario/lista a JSON

```python
book= {
    "title": "La solitudine dei numeri primi",
    "author": "Paolo Giordano",
    "year": "2008",
    "avaiable": True,

}
json_book = json.dumps(book, indent=4)
print(json_book)

print(type(book))
print(type(json_book))
```

### Conversione da JSON a dizionario/lista

```python
json_user ="""[
    {
        "username": "sofia.rossi",
        "email": "sofia.rossi@example.com"
    },
    {
        "username": "giulia.bianchi",
        "email": "giulia.bianchi@example.com"
    }
]"""

users = json.loads(json_user)
print(users)
```

## Gestione di file JSON

```python
book= {
    "title": "La solitudine dei numeri primi",
    "author": "Paolo Giordano",
    "year": "2008",
    "avaiable": True,

}
with open("book.json", "w", encoding="utf-8") as file:
    json.dump(book, file, indent=4, ensure_ascii=False)

```

## Lettura da file JSON

```python
with open("book.json", "r", encoding="utf-8") as file:
    book = json.load(file)

print(book["title"])
```
