# Dizionari

Sono coppie di: chiave-valore scritti tra { } in cui la chiave è di tipo immutabile, mentre il valore pùò essere di qualunque tipologia. La sintassi prevede la scrittura tra parentesi graffe e separazione tra chiave e valore con ":"

`{chiave : valore , chiave1 : valore1 , ...}`

Possono essere creati dizionari vuoti/omogenei o eterogenei.

```python
from asyncio.windows_events import NULL

persona = {"Nome" : "Chiara", "Cognome" : "Scagliola", "eta" : 22}
dizionario_vuoto = {}

print("Dizionario vuoto: ", dizionario_vuoto)

print("Nome persona: ", persona["Nome"])
print("Cognome: ", persona["Cognome"])
print("età: ", persona["eta"])
```

### Aggiunta elementi

É possibile aggiungere delle coppie chiave valore.<br>N.B.: la chiave è univoca, non può essere aggiunta una chiave già esistente.

```python
persona = {"Nome" : "Chiara", "Cognome" : "Scagliola", "eta" : 22}

print("Dizionario originale: ", persona)
print("")

persona["Citta"] = "Torino"
print("Dizionario post aggiunta:", persona)
```

### Modifica elementi

É possibile modificare il valore di una coppia chiave valore.<br>N.B.: la chiave è immutabile e non può essere modificata

```python
persona = {"Nome" : "Chiara", "Cognome" : "Scagliola", "eta" : 22}

print("Dizionario originale: ", persona)
print("")

persona["eta"] = 28
print("Dizionario post modifica:", persona)
```

### Come eliminare degli elementi

Istruzione `del` per eliminare degli elementi (accesso dempre dalla chiave) se non trova la chiave da errore.

```python
persona = {"Nome" : "Chiara", "Cognome" : "Scagliola", "eta" : 22}

print("Dizionario originale: ", persona)
print("")

del persona["eta"]
print("Dizionario post eliminazione di eta: ", persona)
```

## Metodi dei dizionari



## `get()`

Metodo get() restituisce il valore assegnato ad una determinata chiave.

**Sintassi:** `.get("Chiave","Valore di default")` → con questa sintassi posso anche assegnare un valore di default da visualizzare, qualora la chiave non venga trovata.

`get()` restituisce `None` se la chiave è inesistente e non ho inserito un parametro di default.

```python
persona = {"Nome" : "Chiara",
           "Cognome" : "Scagliola",
           "eta" : 22}

print("Nome", persona.get("nome","Mario"))      # setto un parametro di default nel caso in cui la chiave non esista (con ciò posso evitare il KeyError)

print("Cognome", persona.get("cognome"))        # restituisce None se la chiave è inesistente e non ho messo un parametro di default
```

```python
nome = persona.get("nome","Mario")
if nome:
    print("il valore di Nome è:",nome)
else:
    print("non è stato trovato nessun nome associato",nome)
```

## Come rimuovere elementi: metodo `pop()`

```python
persona = {"Nome" : "Chiara",
           "Cognome" : "Scagliola",
           "eta" : 22}
# rimuove la coppia chiave valore specificata e restituisce il valore

age = persona.pop("eta")
print("Valore eliminato: ", age)
print("Dizionario post pop(): ",persona)
```

Se invece vogliamo rimuovere l'ultima coppia chiave valore inserita il metodo è popitem()

```python
persona = {"Nome" : "Chiara",
           "Cognome" : "Scagliola",
           "eta" : 22}

ultimo_valore = persona.popitem()
print("L'ultimo valore è: ", ultimo_valore)
print("Il dizionario dopo la rimozione dell'ultimo valore è: ", persona)
```

## Modifica valori dizionario

`setdefault()` resistituisce il valore di una chiave se esiste, se non esite assegna a quella chiave il valore specificato di default e aggiunge questa coppia al dizionario.


```python
persona = {"Nome" : "Chiara",
           "Cognome" : "Scagliola",
           "eta" : 22}

age = persona.setdefault("Eta", 30)
print("Eta: ", age)
print("Dizionario totale: ", persona)

persona = {"Nome" : "Chiara",
           "Cognome" : "Scagliola",
           "eta" : 22}

age = persona.setdefault("eta", 30)
print("Eta: ", age)
print("Dizionario totale: ", persona)
```

## Come iterare sui dizionari

Esistono tre metodi principali: `key()`, `value()`, `items()`

### Iterazioni sulle chiavi:

Di default l'iterazione è sulle chiavi, se non specificata.

### -`key()`

```python
book = {"Titolo" : "Promessi Sposi", "Autore" : "Alessandro Manzoni", "Lunghezza" : 555}

for key in book:            # default sull'iterazione del dizionario
    print(key, book[key])

for key in book.keys():
    print(key, book[key])
```

### - `values()`

```python
book = {"Titolo" : "Promessi Sposi", "Autore" : "Alessandro Manzoni", "Lunghezza" : 555}

for val in book.values():            # iterazione sui valori
    print(val)

for key in book:            # formula equivalente perchè non stampo le chiavi dopo l'iterazione del dizionario per chiavi
    print(book[key])
```

### - `items()`

```python
book = {"Titolo" : "Promessi Sposi", "Autore" : "Alessandro Manzoni", "Lunghezza" : 555}

for key,value in book.items():
    print(key, value)
```

# Operazioni sui dizionari

`in` verifica la presenza di una chiave nel dizionario. `len()` quante coppie chiave valore sono presenti nel dizionario.

```python
book = {"Titolo" : "Promessi Sposi", "Autore" : "Alessandro Manzoni", "Lunghezza" : 555}

print("Coppie chiave valore presenti: ", len(book))

print("Titolo è una chiave?", "Titolo" in book)     # restituisce True o False

if "Titolo" in book:
    print("Titolo è una chiave.")
```

## Copia dei dizionari

```python
import copy  # libreria da importare per poter usare la funzione deepcopy

book = {"Titolo": "Promessi Sposi",
        "Autore": "Alessandro Manzoni",
        "Lunghezza": 555,
        "Editore": {"Nome": "Rizzoli",
                    "Indirizzo": "Via dei laghi"
                    }
        }
book2 = book  # copia un riferimento in memoria, se modifico uno modifico l'altro
book_copy = book.copy()
book_deep = copy.deepcopy(book)

book["Titolo"] = ["prova"]

print("Libro originale", book)
print("\nCopia assegnazione", book2)  # stesso riferimento di memoria
print("\nCopia superficiale", book_copy)  # .copy() crea una copia indipendente, ma gli oggetti mutabili annidati rimangono condivisi
print("\nCopia profonda", book_deep)  # deepcopuy() crea una copia indipendente ed in questo caso è uguale a copy perchè il valore di Titolo non era una lista o tupla annidata deepcopy

print("\n" * 2 + "*" * 32 + "\n" * 2)

book["Editore"]["Nome"] = "Nuovo"       # è annidato in un altro dizionario, non basta la copia superficiale

print("Libro originale", book)
print("\nCopia assegnazione", book2)  # stesso riferimento di memoria
print("\nCopia superficiale", book_copy)  # stesso riferimento interno
print("\nCopia profonda", book_deep)    #crea una copia indipendente
print("*" * 32)
```

## Operazioni matematiche sui set

Supporta le funzioni insiemistiche di base (unione, intersezione, differenza, differenza simmetrica) consentendo di confrontare due o più set e ottenere un nuovo insieme come risultato.

### Unione

Restituisce tutti gli elementi presenti in almeno uno dei sue set.

```python
insieme_a = {1,2,3,4,5}
insieme_b = {4,5,6,7,8,9}

unione = insieme_a.union(insieme_b)
unione_2 = insieme_a | insieme_b
print(unione)
print(unione_2)
```

### Intersezione

Restituisce solamente gli elementi in comune ai due set, la funzione è ` intersection()` o si utilizza l'operatore `&`.


```python
insieme_a = {1,2,3,4,5}
insieme_b = {4,5,6,7,8,9}

intersezione = insieme_a.intersection(insieme_b)
intersezione_2 = insieme_a & insieme_b
print(intersezione)
print(intersezione_2)
```

### Differenza

Restituisce tutti i valori presenti nel primo set ma non nel secondo set. Si utilizza `difference()` oppure l'operatore `-`.


```python
insieme_a = {1,2,3,4,5}
insieme_b = {4,5,6,7,8,9}

differenza = insieme_a.difference(insieme_b)
differenza_2 = insieme_a - insieme_b

print(differenza)
print(differenza_2)
```

### Differenza simmetrica (UNIONE - INTERSEZIONE)

```python
insieme_a = {1,2,3,4,5}
insieme_b = {4,5,6,7,8,9}

differenza_simm = insieme_a.symmetric_difference(insieme_b)
print(differenza_simm)
```

## Relazioni tra set

Posso verificare relazioni insiemistiche: sottinsieme, sovrainsieme e disgiunzione (mi restituiscono un booleano)

### Sottinsieme
Confronto con l'operatore <= (o `issubset()`) e mi restituisce `True` o False.

```python
insieme_a = {1,2,3,4,5,6,7,8,9}
insieme_b = {1,2,3,4,5}
insieme_c = {9,7,12}

print("Insieme b è sottinsieme di a?", insieme_b <= insieme_a)

print("Insieme a è sottinsieme di c?", insieme_a.issubset(insieme_c))
```

Sottinsieme
Confronto con l'operatore >= (o `issuperset()`) e mi restituisce True o False.

```python
insieme_a = {1,2,3,4,5,6,7,8,9}
insieme_b = {1,2,3,4,5}
insieme_c = {9,7,12}

print("Insieme b è sottinsieme di a?", insieme_b >= insieme_a)

print("Insieme a è sottinsieme di c?", insieme_a.issuperset(insieme_c))
```

### Disgiunti
Insiemi che non han elementi in comune, si usa la funzione `isdisjoint()`


```python
insieme_a = {1,2,3,4,5,6,7,8,9}
insieme_b = {1,2,3,4,5}
insieme_c = {19,17,12}

print("L'insieme b e l'insieme a sono disgiunti? ", insieme_b.isdisjoint(insieme_a))
print("L'insieme b e l'insieme c sono disgiunti? ", insieme_b.isdisjoint(insieme_c))
```
