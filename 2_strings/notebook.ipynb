## Metodi delle stringhe

### Cambio maiuscole/minuscole

```python
name= "Simone"
lower_name = name.lower()
upper_name = name.upper()

print (lower_name)
print(upper_name)
print(name)

title = "crollo in BORSA dell'azienda X."
capitalized_title = title.capitalize()
print("capitalized()", capitalized_title)

first_name = "SIMONE"
capitalized_first_name = first_name.capitalize()
print("capitalized()", capitalized_first_name)

full_name = "SIMONE colOMBa"
titled_full_name = full_name.title()
capitalized_full_name = full_name.capitalize()
print("title()", titled_full_name)
print("capitalize()", capitalized_full_name)
```

## Rimozione spazi

```python
text = "               Testo  con spazi    "
print("Originale:", text)

stripped_text = text.strip()
print("strip():", stripped_text)

lstripped_text = text.lstrip()
print("lstrip():", lstripped_text)

rstripped_text = text.rstrip()
print("rstrip():", rstripped_text)
```

## Sostituzione

```python
text= "A B C D E F"
print("Originale", text)

replaced_text = text.replace(" ", "")
print("replace():", replaced_text)

print()

full_name = "Simone Colomba"
print("Originale:", full_name)

new_name= "Mario"
replaced_full_name = full_name.replace("Simone", new_name)
print("replace():", replaced_full_name)
```

```python
text= "Solitamente programmo in Python. Il Python è un linguaggio molto utile."
print("Originale:", text)

java_text = text.replace("Python", "Java")
print("replace():", java_text)

java_text2 = text.replace("Python", "Java", 1)
print("replace():", java_text2)
```

## Verifica inizio/fine

```python
file_name= "documento.pdf"
is_document=file_name.startswith("doc")
print("E' un documento?", is_document)

is_pdf = file_name.endswith(".pdf")
print("E' un PDF?", is_pdf)
```

## Slicing

```python
text = "Solitamente viaggio in treno."

print("Originale", text)

char = text[23]
print("Carattere con indice 23:", char)

char = text[0]
print("Carattere con indice 0:", char)

char = text[-1]
print("Carattere con indice -1:", char)

part = text[0:5]  ## Oppure [:5]
print("Carattere con indice da 0 a 5:", part)

part = text[23:28]
print("Carattere con indice da 23 a 28:", part)

part = text[20:len(text)] ## Oppure [20:]
print("Carattere con indice da 20 alla fine:", part)

## STEP --> ogni quanto prendere un carattere
part = text[0:len(text):1]
print("Testo con step a 2:", part)

```
