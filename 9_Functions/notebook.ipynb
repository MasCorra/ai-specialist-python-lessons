# Funzioni

Le funzioni permettono di raggruppare una serie di istruzioni e di eseguirle ogni volta che la funzione viene richiamata.

Per definire le istruzioni che la funzione dovrà svolgere si utilizza il comando `def`. Dopo il nome della funzione si scrivono le parentesi tonde e i due punti. Le istruzioni appartenenti alla funzione devono essere indentate.

Il blocco indentato dopo `def` costituisce il corpo della funzione. Definire una funzione non la esegue: le istruzioni vengono eseguite soltanto quando la funzione viene chiamata con le parentesi.

```python
def greet():
    print("=" * 36)
    print("Buongiorno!")
    print("Ti diamo il benvenuto nel programma!")
    print("=" * 36)

greet()
```

Una funzione deve essere definita prima di essere richiamata. Il nome della funzione segue le stesse regole già viste per le variabili e, per convenzione, viene scritto in `snake_case`.


### Docstring

La docstring viene inserita come prima istruzione della funzione e descrive in modo sintetico ciò che la funzione svolge.

## Parametri e argomenti

I **parametri** sono i nomi indicati tra parentesi nella definizione della funzione. Rappresentano i valori che la funzione riceverà.

Gli **argomenti** sono i valori effettivi passati alla funzione quando viene richiamata.

Il numero degli argomenti deve corrispondere ai parametri richiesti, salvo i parametri con un valore predefinito.

Nell'esempio seguente `name`, `age` e `city` sono **parametri**, mentre `"Mario"`, `35` e `"Torino"` sono **argomenti**.

```python
# name, age e city sono parametri delle variabili
def introduzione(name, age, city):
    """
    Introduce l'utente
    :param name: Nome dell'utente
    :param age: Anni dell'utente
    :param city: Città di residenza dell'utente
    """
    print(f"Io sono {name}, ho {age} anni e vivo a {city}.")

# "Mario", 35 e "Torino" sono gli argomenti (valori che assegno ad un parametro)
introduzione("Mario","35","Torino")
```

### Argomenti con parole chiave

Gli argomenti possono essere passati indicando il nome del parametro. In questo caso l'ordine può essere diverso da quello utilizzato nella definizione della funzione.

È possibile usare anche un mix di argomenti posizionali e argomenti con parola chiave (gli argomenti posizionali devono però essere scritti prima di quelli con parola chiave).

```python
def introduce(name, age, city):
    print(f"Io sono {name}, ho {age} anni e vivo a {city}.")

introduce(age=25, name="Mario", city="Torino")

# Posso usare anche un mix dei due metodi per definire i valori dei parametri
introduce("Mario", city="Torino", age=25)

# Quando inizio a passarli con parola chiava, proseguo poi con parola chiave,
# altrimenti viene prodotto un errore:
#
# introduce("Mario", city="Torino", 25)
```

## Parametri con valori predefiniti

Un parametro può avere un valore predefinito, utilizzato quando non viene fornito un argomento corrispondente.

I parametri senza valore predefinito devono essere scritti prima di quelli con valore predefinito, perchè se introduco dei parametri con un valore di default, anche quelli a seguire dovranno averlo.

Forma non valida: `def introduce(name, age=18, city):`

Non è corretta, perché `city` senza valore predefinito si trova dopo `age=18`.

Forma corretta:

Forma corretta:

```python
def introduce(name, age=18, city="Torino"):
    print(f"Io sono {name}, ho {age} anni e vivo a {city}.")

introduce("Mario", 25, "Torino")
introduce("Luca", 19)

# Non restituisce errore, ma il valore "Roma" viene
# assegnato ad age perché è il secondo argomento posizionale
# (Python avvisa del tipo non congruente con quello di default)
introduce("Sofia", "Roma")

# Per modificare soltanto city si usa la parola chiave
introduce("Sofia", city="Roma")
```

### `*args`

`*args` viene usato per definire una funzione della quale non conosco in anticipo il numero di argomenti.

Gli argomenti ricevuti vengono raccolti in una tupla. Il nome `args` è una convenzione: ciò che rende speciale il parametro è l'asterisco `*`.

```python
def sum_all(*numbers):
    print(numbers)
    total = 0
    for n in numbers:
        total += n

    return total

total = sum_all(1,2,3,4,5,6)
print(total)
```

Senza l'asterisco la funzione riceve un solo argomento. In questo caso bisogna passare l'intera lista come unico valore:

```python
def sum_all(numbers):
    print(numbers)
    total = 0
    for n in numbers:
        total += n

    return total

total = sum_all([1,2,3,4,5,6])
print(total)
```

`*args` e un normale parametro che riceve una lista sono diversi. Con `*args` i valori vengono passati separatamente e raccolti in una tupla; con un parametro normale la lista viene passata come unico argomento.

### `*kwargs`

`**kwargs` permette di ricevere un numero non definito di argomenti con parola chiave.

Gli argomenti vengono raccolti in un dizionario: i nomi degli argomenti diventano le chiavi e i valori passati diventano i valori del dizionario.

Anche `kwargs` è un nome convenzionale: sono i due asterischi `**` a raccogliere gli argomenti nel dizionario.

```python
def print_info(**info):
    print(info)

    for key, value in info.items():
        print(key + ":" + value)

print_info(name="Luca", age="30")
```

**RICORDA**:
- `*args` raccoglie gli argomenti posizionali in una tupla
- `**kwargs` raccoglie gli argomenti con parola chiave in un dizionario


## Valori di ritorno

L'istruzione `return` termina la funzione e restituisce un valore al punto in cui la funzione è stata richiamata.

Il valore restituito può essere salvato in una variabile e utilizzato nelle istruzioni successive.

### Restituzione di un solo valore

```python
def multiply(a, b):
    # I parametri sono ciò che inserisco all'interno della funzione
    result = a * b
    result = result * 10
    return result
    # result è ciò che la funzione restituisce come output, non un parametro

func_result = multiply(5, 6)
print(func_result)
```

Se una funzione non contiene `return`, oppure termina senza eseguirlo, restituisce `None`. `print()` mostra un valore, ma non lo restituisce. Inoltre `return` termina immediatamente l’esecuzione della funzione.


```python
def mostra_risultato(a, b):
    print(a + b)


def restituisce_risultato(a, b):
    return a + b


valore_1 = mostra_risultato(2, 3)
valore_2 = restituisce_risultato(2, 3)
print("Valore restituito dal primo esempio:", valore_1)
print("Valore restituito dal secondo esempio:", valore_2)
```

### Restituzione di più valori

Una funzione può restituire più valori separandoli con una virgola. Python li raccoglie in una tupla.

È possibile accedere ai singoli risultati tramite indice oppure usare l'unpacking.

```python
def calculate(a, b):
    sum_result = a + b
    sub_result = a - b
    return sum_result, sub_result

results = calculate (10,5)
print(results)

sum_result = results[0]
print(sum_result)
sub_result = results[1]
print(sub_result)

# Con unpacking
sum_result, sub_result = calculate(10, 5)

for result in results:
    print(result)
```

Quando si restituiscono più valori separati da virgole, Python crea una tupla. I valori possono essere letti tramite indice oppure assegnati direttamente con l’unpacking, già utilizzato nelle lezioni precedenti.


## Scope delle variabili

Lo **scope** indica la parte del programma in cui una variabile è disponibile.

- Una variabile definita fuori dalle funzioni ha **scope globale**.
- Una variabile definita dentro una funzione ha **scope locale** ed è disponibile soltanto all'interno di quella funzione.

Una funzione può leggere una variabile globale se al suo interno non viene definita una variabile locale con lo stesso nome.

```python
# Variabile con scope globale

app = "Programma di prova"

def welcome():
    # Variabile con scope locale
    day = "venerdì"
    print(f"Benvenuto nell'app \"{app}\"! Oggi è {day}!")

welcome()

# day non è disponibile fuori dalla funzione,
# perché è una variabile locale ed esiste solo dentro quella funzione
# print(day)
```

Un parametro della funzione è una variabile locale. Se ha lo stesso nome di una variabile globale, all'interno della funzione viene utilizzato il parametro.

```python
# Variabile con scope globale
app = "Programma di prova"

def welcome(app):
    # Il parametro locale app è più importante e sovrascrive quella esterna: viene utilizzato al posto della variabile globale
    print(f"Benvenuto nell'app \"{app}\"!")

welcome("TEST")
```

Assegnando un valore a `app` dentro la funzione viene creata una nuova variabile locale. La variabile globale con lo stesso nome non viene modificata.

Le variabili locali esistono durante l'esecuzione della funzione e non sono disponibili fuori da essa.

```python
app = "Programma di prova"

def welcome():
    # Crea una nuova variabile app con scope locale, non ridefinisce quella esterna
    app = "Nuovo Programma"
    print(f"Benvenuto nell'app \"{app}\"!")

# le variabili all'interno delle funzioni (variabili locali) una volta terminata la funzione vengono ripristinate.

welcome()
print(app)
```

## Accesso alle variabili globali

La parola chiave `global` permette di indicare che, dentro una funzione, si vuole utilizzare e modificare una variabile globale invece di crearne una locale.

Se la funzione deve soltanto leggere una variabile globale, `global` non è necessario. `global` serve soltanto quando nella funzione è necessario riassegnare una variabile globale.

```python
count = 0

def greet():
    global count
    # porta la variabile count all'interno della def della funzione
    # da adesso può essere utilizzata e modificata nella def
    print("Ciao")

    count = count + 1
    print(f"Oggi la funzione è stata richiamata {count} volte. ")

greet()
greet()
greet()
greet()
```
