# DEFINIZIONE DI UNA CLASSE

**Classe** = stampino del biscotto, **oggetto** = il biscotto, **attributi** = caratteristiche del biscotto

**Classe:** modello da cui si creano oggetti. **Istanza (oggetto):** elemento concreto della classe.

**Attributo:** dato associato a un oggetto. **`__init__`:** è il costruttore e inizializza l’istanza; **`self`** indica l’oggetto corrente.

```python
class Person:
  def __init__(self,first_name,surname, age = 20): #metodo che viene richiamato quando creo la persona
      self.first_name = first_name #memorizza il nome per quell'oggetto
      self.surname = surname
      self.name = first_name + ' ' + surname
      self.age = age

chiara = Person("Chiara","Scagliola",27)
lucia = Person('Lucia','Gialli',32)
print(chiara.name)
print(lucia.name, lucia.age)

print('tipo', type(chiara))
```

**Metodo di istanza:** funzione definita in una classe che opera su un oggetto e riceve `self` come primo parametro.

```python
class Person:
  def __init__(self,first_name,surname, age = 20): #metodo che viene richiamato quando creo la persona
      self.first_name = first_name #memorizza il nome per quell'oggetto
      self.surname = surname
      self.name = first_name + ' ' + surname
      self.age = age
      self.adult = self.age > 18

  def greet(self):
    print(f'Ciao! sono {self.name} e ho {self.age} anni!')

chiara = Person("Chiara","Scagliola",27)
chiara.greet()

```

**Incapsulamento:** organizzare dati e metodi all’interno della classe.

**`_attributo`:** uso interno per convenzione. **`__attributo`:** attiva il *name mangling* (rinomina interna per ridurre gli accessi accidentali, non una vera protezione privata).

```python
class Account:
  def __init__(self, username,email,pin):
    self.username = username
    self.email = email
    self._pin = pin # sconsigliato accesso dall'esterno, ma comunque possibile
    self.__secret_code = username[:3] # ci puoi accedere solo dall'interno

  def verify_pin (self, pin ):
    return self._pin == pin

  def describe(self):
    print(f'account di {self.username}, email: {self.email}')

#voglio creare l'oggetto account
account1 = Account('chiara.s','chiara.scagliola99@gmail.com','1234')
#verifica del pin
print(account1.verify_pin('123567'))
print(account1.verify_pin('1234'))
account1.describe()
```

**Composizione:** un oggetto contiene o utilizza altri oggetti come parte del proprio funzionamento.

```python
class BankLog:
  def __init__(self, old_balance, new_balance , operation):
    self.old_balance = old_balance
    self.new_balance = new_balance
    self.operation = operation


  def summary(self):
    print(f'[{self.operation}] {self.old_balance}-> {self.new_balance}')

class BankAccount:
  def __init__(self, first_name,last_name, email):
    self.first_name = first_name
    self.last_name = last_name
    self.email = email
    self.__balance = 0
    self.__logs = []

  def show(self):
    print(f'il conto corrente è di :{self.first_name} {self.last_name}')
    print('il saldo è di:', self.__balance)
    for log in self.__logs:
      log.summary()


  def deposit(self, amount):
    old_balance = self.__balance
    self.__balance += amount
    self.__logs.append(BankLog(old_balance,self.__balance, '+'))

  def withdrow(self, amount):
    old_balance = self.__balance
    self.__balance -= amount
    self.__logs.append(BankLog(old_balance,self.__balance, '-'))

account = BankAccount('Chiara', 'Scagliola', 'chiara.sca@gmail.com')

account.show()
account.deposit(1000)
account.withdrow(50)
account.show()
```

**Attributo di classe:** appartiene alla classe ed è condiviso tra le istanze, salvo ridefinizioni.

**`@classmethod`:** riceve `cls`, cioè la classe. **`@staticmethod`:** non riceve automaticamente né `self` né `cls`.

```python
class BankAccount:
  bank_name = 'Intesa San Paolo'
  total_accounts = 0

  def __init__(self, first_name,last_name, email):
    self.first_name = first_name
    self.last_name = last_name
    self.email = email
    self.__balance = 0
    BankAccount.total_accounts +=1
  def show(self):
      print(f'il conto corrente è di :{self.first_name} {self.last_name}')
      print('il saldo è di:', self.__balance)
      print('Account Totali:', self.total_accounts)
  def deposit(self, amount):
    old_balance = self.__balance
    self.__balance += amount


  def withdrow(self, amount):
    old_balance = self.__balance
    self.__balance -= amount

  @classmethod
  def show_numer_accounts(cls):
    print('il numero di account registrati è: ', cls.total_accounts)

  @staticmethod
  def is_amount_valid(amount):
    return 0<amount<1000

account1 = BankAccount('Chiara','scagliola','chiara.scagliola@gmail.com')
account1.show()
print('--------------------------------------------------')
account2 = BankAccount('Lucia','Ferni','fernilucia@hotmail.it')
account1.show()
account2.show()

print('-----------------------------------------------------')
print(BankAccount.total_accounts)
BankAccount.show_numer_accounts()
account1.show_numer_accounts()

BankAccount.is_amount_valid(100)
```
