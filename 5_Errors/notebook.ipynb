## Cattura errori specifici

```python
try:
    number= int(input("Inserisci un numero: "))
    print("Il numero inserito è:", number)
    print("somma", 10 + number)
    print("divisione", 100/number)

except ValueError:
    number = 5   #Valore assegnato nel momento in cui si finisce nell'eccezione
    print("errore di valore: il valore inserito non è valido!")
except ZeroDivisionError:
    print("Stai dividendo per zero!")
```

## Cattura Errori

```python
try:
    number= int(input("Inserisci un numero: "))
    print("Il numero inserito è:", number)
    print("somma", 10 + number)

except ValueError as error:
    number = 5   #Valore assegnato nel momento in cui si finisce nell'eccezione
    print("errore di valore: il valore inserito non è valido!")
    print(error)
```

## Else

```python
try:
    number= int(input("Inserisci un numero: "))
    print("Il numero inserito è:", number)
    print("somma", 10 + number)

except ValueError as error:
    number = 5   #Valore assegnato nel momento in cui si finisce nell'eccezione
    print("errore di valore: il valore inserito non è valido!")
    print(error)
else:
    print("Tutto è andato bene!")
    print("la somma del tuo valore e 10 è:", number + 10)
```

## Finally

```python
try:
    number= int(input("Inserisci un numero: "))
    print("Il numero inserito è:", number)
    print("somma", 10 + number)

except ValueError as error:
    number = 5   #Valore assegnato nel momento in cui si finisce nell'eccezione
    print("errore di valore: il valore inserito non è valido!")
    print(error)
finally:
    print("Questo pezzo di codice viene eseguito in ogni caso")
```

```python
try:
    number= int(input("Inserisci un numero: "))

    result = 200/number

except ValueError as error:
    print("errore di valore: il valore inserito non è valido!")
    print(error)
except ZeroDivisionError:
    print("Stai dividendo per zero!")
else:
    print("Il risultato della divisione è:", result)
finally:
    print("Operazione completata")
```
