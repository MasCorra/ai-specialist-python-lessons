## Variabili

**Convenzione per i nomi**: i nomi delle variabili devono essere scritti in snake_case (tutto minuscolo e con i _ al posto degli spazi). Si possono utilizzare solo caratteri alfanumerici (lettere e numeri). Non si puo cominciare il nome di una variabile con un numero.

```python
# Variabili di tipo string (str)
name1 = "Simone"
name2 = "Sofia"

# Variabili di tipo integer (int)
age = 23

# Variabile di tipo float
height = 1.84

# Variabile di tipo boolean (bool)
adult = True
```

### Convertire il tipo di una variabile

```python
current_year_text = "2026"

# Converte la variabile da str a int
# Nel caso ci fossero dei caratteri non validi, darebbe errore
current_year = int(current_year_text)

height = 1.84

int_height=int(height)
print(int_height)

price_text = "1.50"

price=float(price_text)
print(price)
```

## Operatori aritmetici

```python
x = 10
y = 2

# Somma x e y
result1 = x + y
print("Somma: " + str(result1))

# Sottrae y da x
result2 = x - y

#Moltiplica x per y
result3 = x * y

# Divide x per y
result4 = x / y

#Eleva x alla y
result5 = x ** y

#Calcola il resto della divisione di x per y
result6 = x % y

#Calcola la parte intera della divisione di x per y
result7 = x // y
```
